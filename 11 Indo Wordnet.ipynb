```python
import pyiwn
```

```python
iwn = pyiwn.IndoWordNet(lang=pyiwn.Language.MARATHI)
```

```python
word = iwn.synsets('धरती')
```

```python
word
```

```python
# Explore the specific synsets
```

```python
word[0].synset_id()
```

```python
word[0].pos()
```

```python
# meaning
word[0].gloss()
```

```python
# Example in senetence
word[0].examples()
```

```python
# synonyms
word[0].lemma_names()
```

```python
# Hyponyms
```

```python
hyponyms = iwn.synset_relation(word[0], pyiwn.SynsetRelations.HYPONYMY)
```

```python
for h in hyponyms: 
    print(h.lemma_names()[0])
```

```python
word = iwn.synsets('आंबा')
hyponyms = iwn.synset_relation(word[0], pyiwn.SynsetRelations.HYPONYMY)
for h in hyponyms: 
    print(h.lemma_names()[0])
```
