#### PyTTS

```python
import pyttsx3
```

```python
engine = pyttsx3.init()
```

```python
# Set the properties
engine.setProperty('rate', 100)
engine.setProperty('volume', 0.9)
```

```python
text = 'Hello Welcome to the world of Python Programming.'
```

```python
engine.say(text)
engine.runAndWait()
```

#### Google TTS

```python
from gtts import gTTS
from playsound import playsound
```

```python
# create the object of TTS
tts = gTTS(text='Welcome to Python Programming.', lang = 'en')
```

```python
# save the audio
tts.save('audio.mp3')
playsound('audio.mp3')
```

```python
# create the object of TTS
tts = gTTS(text='पायथॉन प्रोग्रॅमिंगमध्ये तुमचे स्वागत आहे.', lang = 'mr')
```

```python
# save the audio
tts.save('audio.mp3')
playsound('audio.mp3')
```

```python
# create the object of TTS
tts = gTTS(text='Welcome to Python Programming.', lang = 'en', tld='co.uk')
tts.save('audio.mp3')
playsound('audio.mp3')
```
