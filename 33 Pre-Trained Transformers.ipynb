#### Question-Answering

```python
qa = pipeline('question-answering')
```

```python
text = 'New Delhi is capital city of India.'
question = 'What is capital city of India?'
result_qa = qa(question=question, context=text)
df = pd.DataFrame([result_qa])
df
```
