```python
chess = np.zeros([8,8])
for i in range(8): 
    for j in range(8): 
        if (i+j) % 2 == 0:
            chess[i,j] = 1
```

#### RGB Channels

```python
r = img_plt[:,:,0]
```

```python
r
```

```python
g = img_plt[:,:,1]
b = img_plt[:,:,2]
```

```python
plt.figure(figsize=(10,3))
plt.subplot(1,3,1)
plt.imshow(r, cmap = 'gray')
plt.subplot(1,3,2)
plt.imshow(g, cmap = 'gray')
plt.subplot(1,3,3)
plt.imshow(b, cmap = 'gray')
```

```python
r.shape, g.shape, b.shape
```

```python
img_avg = (r + g + b) // 3
```

```python
plt.imshow(img_avg, cmap = 'gray')
```
